### Folded SFS
First I calculate the folded SFS for each population. I choses the folded SFS because I don't have an outgroup genome. 

```bash
# make bamlist 
BAMDIR="/media/ssd/popglen_workflow/results/datasets/p_arcticus_02_2026/bamlists"
OUTDIR="/media/ssd/Bioinformatics/p_arcticus_02/bamlist/SFS_populations/Xavier_pop"

cat \
"$BAMDIR/p_arcticus_02_2026.dataset-ref_California.bamlist" \
"$BAMDIR/p_arcticus_02_2026.dataset-ref_Washington.bamlist" \
"$BAMDIR/p_arcticus_02_2026.dataset-ref_Oregon.bamlist" \
> "$OUTDIR/West.bamlist"

cat \
"$BAMDIR/p_arcticus_02_2026.dataset-ref_Michigan.bamlist" \
"$BAMDIR/p_arcticus_02_2026.dataset-ref_Minnesota.bamlist" \
> "$OUTDIR/South_central.bamlist"

cat \
"$BAMDIR/p_arcticus_02_2026.dataset-ref_Manitoba.bamlist" \
"$BAMDIR/p_arcticus_02_2026.dataset-ref_Alberta.bamlist" \
> "$OUTDIR/North_central.bamlist"

# Remove MB19073 from North_central
grep -v 'MB19073' "$OUTDIR/North_central.bamlist" > "$OUTDIR/North_central.tmp"
mv "$OUTDIR/North_central.tmp" "$OUTDIR/North_central.bamlist"

# Add MB19073 to South_central
grep 'MB19073' "$BAMDIR/p_arcticus_02_2026.dataset-ref_Manitoba.bamlist" >> "$OUTDIR/South_central.bamlist"

cp \
"$BAMDIR/p_arcticus_02_2026.dataset-ref_NewMexico.bamlist" \
"$OUTDIR/NewYork.bamlist"
#there is a mistae in bamlist names and the NewYork one is called NewMexico 

cp \
"$BAMDIR/p_arcticus_02_2026.dataset-ref_Quebec.bamlist" \
"$OUTDIR/Quebec.bamlist"


#Now the bamlist files are different since it changed directory so i am changin it 
for file in "$OUTDIR"/*.bamlist; do
    sed -i 's|/media/laurie/5A9848239847FFCD/|/media/ssd/|g' "$file"
done
```

```bash
DIR="/media/ssd/Bioinformatics/p_arcticus_02/bamlist/SFS_populations/Xavier_pop"
SNPlist=/media/ssd/Bioinformatics/ngsLD/LD_Pruning/LDpruned_snps.list
REF=/media/ssd/Raw_data/ncbi_tridactylus_ref_genome/ncbi_dataset/data/GCA_045782225.1/GCA_045782225.1_ASM4578222v1_genomic.fna
OUTDIR=/media/ssd/Bioinformatics/p_arcticus_02/downstream_analysis/SFS/outputs/SNPpruned/Xavier_pop

for POP in West South_central North_central NewYork Quebec; do     echo $POP;      angsd -b $DIR/${POP}.bamlist           -ref $REF           -anc $REF      -sites "$SNPlist"     -out $OUTDIR/$POP          -minMapQ 30 -minQ 30 -minInd 2 -setMinDepthInd 1           -GL 2 -doSaf 1; done
```
